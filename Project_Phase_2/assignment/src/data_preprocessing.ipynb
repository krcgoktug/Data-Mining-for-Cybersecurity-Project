```python
import pandas as pd
import numpy as np
import glob

# 1. MERGING FILES
print("Merging data...")
# List all csv files in the folder
all_files = glob.glob("*.csv")
li = []

for filename in all_files:
    df_temp = pd.read_csv(filename, index_col=None, header=0, encoding='cp1252')
    li.append(df_temp)

# Combine everything into a single table (dataframe)
df = pd.concat(li, axis=0, ignore_index=True)
print(f"Total {len(df)} rows of data loaded.")

# 2. DATA CLEANING
# Clean spaces in column names
df.columns = df.columns.str.strip()

# Delete missing (NaN) values
df.dropna(inplace=True)

# Detect and clean infinite (inf) values
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)

print("Missing and infinite values cleaned.")
print("First 5 rows of data:")
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Label'] = le.fit_transform(df['Label'])

print("Conversion to numbers completed!")
# To see which number corresponds to which class:
print(dict(zip(le.classes_, le.transform(le.classes_))))
```
