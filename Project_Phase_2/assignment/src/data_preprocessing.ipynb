```python
import pandas as pd
import numpy as np
import glob
from sklearn.preprocessing import LabelEncoder

# --- STEP 1: DATA INGESTION AND MERGING ---
# We are merging multiple daily network traffic logs (Monday to Friday) into a single unified dataframe.
# This creates a comprehensive dataset for a more robust cybersecurity analysis, allowing us to see weekly patterns.
print("Merging datasets from Monday to Friday...")
all_files = glob.glob("*.csv")
li = []

for filename in all_files:
    # Each CSV is read with 'cp1252' encoding to prevent character errors during data ingestion.
    df_temp = pd.read_csv(filename, index_col=None, header=0, encoding='cp1252')
    li.append(df_temp)

# Combining all separate daily files into one master table.
df = pd.concat(li, axis=0, ignore_index=True)
print(f"Total {len(df)} rows of raw data loaded successfully.")

# --- STEP 2: DATA CLEANING AND STANDARDIZATION ---
# We strip any leading or trailing white spaces from column names.
# This is a critical step to prevent "KeyError" when referencing features like 'Label' or 'Flow Duration'.
df.columns = df.columns.str.strip()

# Missing values (NaN) are removed to maintain the integrity of the feature set.
# Incomplete data records can lead to skewed results and bias during the machine learning training process.
df.dropna(inplace=True)

# Cybersecurity datasets often contain infinite (inf) values, usually caused by zero-division in 'Flow Duration' calculations.
# These infinite values must be replaced with NaN and removed, otherwise they will cause the ML model to crash during training.
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)

print("Data cleaning completed: Missing and infinite values have been removed.")

# --- STEP 3: CATEGORICAL TO NUMERICAL TRANSFORMATION ---
# In cybersecurity datasets, the 'Label' column often contains text (e.g., 'DDoS', 'Benign').
# Machine learning models require numerical input to function. We use LabelEncoder to 
# transform these categories into distinct integer values, enabling mathematical processing.
le = LabelEncoder()
df['Label'] = le.fit_transform(df['Label'])

print("Feature encoding completed successfully!")
# Displaying the class mapping so other team members can identify which number 
# corresponds to which attack type during the model evaluation phase.
print("Encoded Label Mapping:")
print(dict(zip(le.classes_, le.transform(le.classes_))))

# --- STEP 4: FINAL DATA EXPORT ---
# We are preparing the final, cleaned version of the dataset for project Phase 3.
# We save the file as 'clean_cyber_data.csv' to maintain naming consistency across all project files and instructions.
print("Preparing the finalized file, please wait...")
df.to_csv('clean_cyber_data.csv', index=False)

# NOTE: We use standard pandas to_csv for general compatibility, ensuring the code 
# works on any local environment (VS Code, Jupyter) as well as cloud platforms without errors.
print("SUCCESS: File saved as 'clean_cyber_data.csv'.")
print("You can now find the processed dataset in your project folder.")
```
